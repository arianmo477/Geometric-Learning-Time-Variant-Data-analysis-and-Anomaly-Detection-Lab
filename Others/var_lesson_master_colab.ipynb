## 1. Motivation

A **univariate** autoregressive model only explains one variable using its own past. For many real systems, this is too restrictive.

Examples:
- GDP growth and interest rates influence each other;
- inflation and unemployment interact dynamically;
- multiple sensor signals evolve jointly.

This motivates **multivariate time series models**.

### Definition: Multivariate time series
A multivariate time series is a sequence

$$
\{\mathbf y_t\}_{t \in \mathbb Z}, \qquad \mathbf y_t \in \mathbb R^k,
$$

where each component of $\mathbf y_t$ is one observed variable at time $t$.

Example with $k=2$:

$$
\mathbf y_t =
\begin{pmatrix}
\text{GDP}_t \\
\text{InterestRate}_t
\end{pmatrix}.
$$



---
## 2. Definition of a VAR($p$)

### Definition
A $k$-dimensional process $\{\mathbf y_t\}$ follows a **Vector Autoregression of order $p$**, written VAR($p$), if

$$
\mathbf y_t = \mathbf c + A_1 \mathbf y_{t-1} + A_2 \mathbf y_{t-2} + \cdots + A_p \mathbf y_{t-p} + \boldsymbol\varepsilon_t,
$$

where:
- $\mathbf c \in \mathbb R^k$ is a constant vector;
- $A_1, \dots, A_p \in \mathbb R^{k \times k}$ are coefficient matrices;
- $\boldsymbol\varepsilon_t$ is a white-noise vector with
  $$E[\boldsymbol\varepsilon_t] = 0, \qquad \operatorname{Cov}(\boldsymbol\varepsilon_t)=\Sigma.$$

### Interpretation
Each variable depends on:
- its **own lagged values**, and
- the **lagged values of all the other variables**.

### Example: VAR(1) with two variables

$$
\begin{cases}
 y_{1,t} = c_1 + a_{11} y_{1,t-1} + a_{12} y_{2,t-1} + \varepsilon_{1,t}, \\
 y_{2,t} = c_2 + a_{21} y_{1,t-1} + a_{22} y_{2,t-1} + \varepsilon_{2,t}.
\end{cases}
$$

Here:
- $a_{12}$ measures the lagged effect of variable 2 on variable 1;
- $a_{21}$ measures the lagged effect of variable 1 on variable 2.



---
## 3. Companion form — full derivation

The companion form rewrites a VAR($p$) as a VAR(1) in a higher-dimensional state vector.

Start from

$$
\mathbf y_t = A_1\mathbf y_{t-1} + A_2\mathbf y_{t-2} + \cdots + A_p\mathbf y_{t-p} + \boldsymbol\varepsilon_t.
$$

For simplicity, the constant is omitted in the algebra below. It can always be added back afterward.

### Step 1: Build the stacked state vector
Define

$$
\mathbf Y_t =
\begin{pmatrix}
\mathbf y_t \\
\mathbf y_{t-1} \\
\vdots \\
\mathbf y_{t-p+1}
\end{pmatrix}
\in \mathbb R^{kp}.
$$

This vector contains the current observation and the previous $p-1$ lags.

### Step 2: Express the first block
The first block of $\mathbf Y_t$ is exactly the original VAR equation:

$$
\mathbf y_t = A_1\mathbf y_{t-1} + A_2\mathbf y_{t-2} + \cdots + A_p\mathbf y_{t-p} + \boldsymbol\varepsilon_t.
$$

### Step 3: Express the remaining blocks
The second block of $\mathbf Y_t$ is simply

$$
\mathbf y_{t-1} = I_k\mathbf y_{t-1},
$$

which comes from shifting the lag vector down by one step. Likewise,

$$
\mathbf y_{t-2} = I_k\mathbf y_{t-2}, \qquad \dots \qquad \mathbf y_{t-p+1}=I_k\mathbf y_{t-p+1}.
$$

### Step 4: Write the matrix system
Combining the blocks gives

$$
\mathbf Y_t = F\mathbf Y_{t-1} + \mathbf u_t,
$$

where the **companion matrix** is

$$
F =
\begin{pmatrix}
A_1 & A_2 & \cdots & A_{p-1} & A_p \\
I_k & 0   & \cdots & 0       & 0 \\
0   & I_k & \cdots & 0       & 0 \\
\vdots & \vdots & \ddots & \vdots & \vdots \\
0 & 0 & \cdots & I_k & 0
\end{pmatrix},
$$

and the innovation vector is

$$
\mathbf u_t =
\begin{pmatrix}
\boldsymbol\varepsilon_t \\
0 \\
\vdots \\
0
\end{pmatrix}.
$$

### Statement
Therefore, every VAR($p$) can be rewritten as a VAR(1) in dimension $kp$.

This form is extremely useful because many theoretical results reduce to studying the single matrix $F$.

## 4. Stability and stationarity

### Definition: Weak stationarity
A multivariate process $\{\mathbf y_t\}$ is **weakly stationary** if:

1. $E[\mathbf y_t]=\boldsymbol\mu$ is constant over time;
2. $\operatorname{Cov}(\mathbf y_t, \mathbf y_{t-h})$ depends only on the lag $h$.

### Stability condition for VAR($p$)
A VAR($p$) is stable if

$$
\det\left(I_k - A_1 z - A_2 z^2 - \cdots - A_p z^p\right) \neq 0
\quad \text{for all } |z| \le 1.
$$

### Equivalent companion-matrix statement
A VAR($p$) is stable if and only if **all eigenvalues of the companion matrix $F$ have modulus strictly smaller than 1**:

$$
|\lambda_i(F)| < 1 \quad \text{for all } i.
$$

### Interpretation
- If the eigenvalues are inside the unit circle, shocks eventually die out.
- If at least one eigenvalue lies on or outside the unit circle, the process is nonstationary or explosive.

### Consequence: MA($\infty$) representation
If the VAR is stable, then it admits a convergent infinite moving-average representation:

$$
\mathbf y_t = \boldsymbol\mu + \sum_{h=0}^{\infty} \Phi_h \boldsymbol\varepsilon_{t-h}.
$$

This representation is the basis for **impulse response functions**.

## 5. Numerical example — complete step by step

Consider the VAR(1)

$$
\begin{cases}
 y_{1,t} = 0.5y_{1,t-1} + 0.2y_{2,t-1} + \varepsilon_{1,t}, \\
 y_{2,t} = 0.1y_{1,t-1} + 0.4y_{2,t-1} + \varepsilon_{2,t}.
\end{cases}
$$

In matrix form,

$$
\mathbf y_t = A\mathbf y_{t-1} + \boldsymbol\varepsilon_t,
\qquad
A = \begin{pmatrix}0.5 & 0.2 \\ 0.1 & 0.4\end{pmatrix}.
$$

### Goal
We will:
1. compute the eigenvalues of $A$;
2. verify stability;
3. examine how a shock propagates through time.


```python

import numpy as np

A = np.array([[0.5, 0.2],
              [0.1, 0.4]])

eigvals = np.linalg.eigvals(A)
A, eigvals
```


### Characteristic polynomial
For a $2\times2$ matrix, the characteristic equation is

$$
\det(A - \lambda I)=0.
$$

Here,

$$
\det\begin{pmatrix}
0.5-\lambda & 0.2 \\
0.1 & 0.4-\lambda
\end{pmatrix}=0.
$$

Hence,

$$
(0.5-\lambda)(0.4-\lambda)-0.02=0,
$$

which simplifies to

$$
\lambda^2 - 0.9\lambda + 0.18=0.
$$

The roots are

$$
\lambda_1 = 0.6, \qquad \lambda_2 = 0.3.
$$

Since both are strictly smaller than 1 in modulus, the system is **stable**.


```python

# Verify powers of A decay
for h in range(1, 6):
    print(f"A^{h} =
{np.linalg.matrix_power(A, h)}
")
```


### Impulse propagation
Suppose there is a one-unit shock in the first equation at time $t$:

$$
\boldsymbol\varepsilon_t = \begin{pmatrix}1 \\ 0\end{pmatrix}.
$$

At horizon 1, the effect is

$$
A\begin{pmatrix}1 \\ 0\end{pmatrix} =
\begin{pmatrix}0.5 \\ 0.1\end{pmatrix}.
$$

So the shock affects both variables after one period.

At horizon 2, the effect is

$$
A^2\begin{pmatrix}1 \\ 0\end{pmatrix}.
$$

The repeated multiplication by $A$ captures how the shock propagates through the dynamic system.


```python

shock = np.array([1.0, 0.0])
for h in range(6):
    effect = np.linalg.matrix_power(A, h) @ shock
    print(f"Horizon {h}: {effect}")
```

## 6. Simulating a VAR(1) series

The next cell simulates data from the same VAR(1) system. This is useful both for teaching and for empirical practice.


```python

import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
T = 300
Sigma = np.array([[1.0, 0.3],
                  [0.3, 1.0]])

errors = np.random.multivariate_normal(mean=[0, 0], cov=Sigma, size=T)
Y = np.zeros((T, 2))

for t in range(1, T):
    Y[t] = A @ Y[t-1] + errors[t]

sim_data = pd.DataFrame(Y, columns=["y1", "y2"])
sim_data.head()
```

```python

sim_data.plot(figsize=(10, 4), title="Simulated VAR(1) series")
plt.show()
```


---
## 7. Estimating a VAR in Python

A useful practical fact is that each equation in a VAR can be estimated by OLS because the regressors are the same across equations.

In Python, `statsmodels` provides a direct implementation.


```python

# If needed in Colab, uncomment the next line:
# !pip -q install statsmodels

from statsmodels.tsa.api import VAR

model = VAR(sim_data)
results = model.fit(maxlags=4, ic='aic')

print(results.summary())
```


### Statement: Lag order selection
In practice, the lag length $p$ is often selected using information criteria such as AIC or BIC.

The trade-off is standard:
- too few lags: missing important dynamics;
- too many lags: overfitting and inefficient estimation.


```python

print("Selected lag order:", results.k_ar)
print("Estimated coefficient matrices:")
for i, coef in enumerate(results.coefs, start=1):
    print(f"
A_{i} =
{coef}")
```

## 8. Impulse Response Functions (IRFs)

### Definition
The **impulse response function** measures the response of variable $i$ at horizon $h$ to a one-unit shock in variable $j$ at time $t$:

$$
IRF_{ij}(h) = \frac{\partial y_{i,t+h}}{\partial \varepsilon_{j,t}}.
$$

### Interpretation
IRFs answer questions such as:
- What happens to GDP after a shock to the interest rate?
- How long does the effect persist?
- Does the effect change sign over time?


```python

irf = results.irf(12)
irf.plot()
plt.show()
```


### Note on orthogonalization
If the components of $\boldsymbol\varepsilon_t$ are correlated, then a "shock to one variable" is not uniquely defined.
A common solution is **Cholesky orthogonalization**, which transforms the innovations into uncorrelated shocks.

This is why ordering may matter in standard orthogonalized IRFs.



---
## 9. Granger causality

### Definition
A variable $x_t$ **Granger-causes** $y_t$ if past values of $x_t$ improve the prediction of $y_t$ beyond the information already contained in past values of $y_t$.

### Important remark
Granger causality is **predictive**, not necessarily structural or philosophical causality.


```python

# Test whether y2 helps predict y1
# Null hypothesis: y2 does NOT Granger-cause y1
print(results.test_causality('y1', ['y2'], kind='f'))
```


---
## 10. Forecast Error Variance Decomposition (FEVD)

### Definition
The **forecast error variance decomposition** tells us what proportion of the forecast error variance of each variable is explained by shocks to each variable at different horizons.

This complements IRFs:
- IRFs show the **direction and path** of responses;
- FEVD shows the **relative importance** of each shock.


```python

fevd = results.fevd(12)
fevd.summary()
```

```python

fevd.plot()
plt.show()
```

## 12. Final recap

### Main definitions and statements to remember

- A **VAR($p$)** models a vector of time series using lagged values of **all variables**.
- The **companion form** rewrites VAR($p$) as VAR(1) in dimension $kp$.
- A VAR is **stable** if all eigenvalues of the companion matrix lie **inside the unit circle**.
- If stable, the VAR has an **MA($\infty$)** representation.
- **IRFs** track the effect of a shock over time.
- **FEVD** quantifies the relative contribution of each shock to forecast uncertainty.
- **Granger causality** is about predictive content, not necessarily causal mechanism.
